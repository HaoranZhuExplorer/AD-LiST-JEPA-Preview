```python
import json
import numpy as np
import matplotlib.pyplot as plt

base_dir = "<your_path>/AD-LiST-JEPA/output/waymo_models/network_bev_ocf"

exp1_name = "bev_ocf_ad_list_jepa_1206_sampled-interval-2"
exp2_name = "bev_ocf_scratch_1206_sampled-interval-2"

epochs = range(1, 11)
fg_acc_exp1 = []
fg_acc_exp2 = []

# ---- Load Experiment 1 ----
for e in epochs:
    file_path = f"{base_dir}/{exp1_name}/eval/epoch_{e}/eval_ssl/bev_ocf_eval_results.json"
    with open(file_path, "r") as f:
        data = json.load(f)
    fg_acc_exp1.append(data[f"epoch_{e}"]["fg_acc"])

# ---- Load Experiment 2 ----
for e in epochs:
    file_path = f"{base_dir}/{exp2_name}/eval/epoch_{e}/eval_ssl/bev_ocf_eval_results.json"
    with open(file_path, "r") as f:
        data = json.load(f)
    fg_acc_exp2.append(data[f"epoch_{e}"]["fg_acc"])

# ---- Plot ----
epochs = np.arange(1, 11)

plt.figure(figsize=(8,5))
plt.plot(epochs, fg_acc_exp1, marker='o', label="pretrained with AD-LiST-JEPA")
plt.plot(epochs, fg_acc_exp2, marker='s', label="trained from scratch")

plt.xlabel("Epoch")
plt.ylabel("Foreground Accuracy")
plt.title("Comparison of Foreground Test Accuracy Across Epochs (1–10)")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
